```python
# Set creation
s = {10, 20, 30, 40}

# Display set
print("Initial set:", s)

# Add element
s.add(50)
print("After add:", s)

# Update (add multiple elements)
s.update([60, 70])
print("After update:", s)

# Remove element
s.remove(20)
print("After remove:", s)

# Discard element (no error if not present)
s.discard(100)

# Pop (removes random element)
s.pop()
print("After pop:", s)

# Length of set
print("Length:", len(s))

# Membership
print("Is 30 in set?", 30 in s)

# Set operations
a = {1, 2, 3}
b = {3, 4, 5}

print("Union:", a | b)
print("Intersection:", a & b)
print("Difference (a - b):", a - b)

# Clear set
s.clear()
print("After clear:", s)
```
