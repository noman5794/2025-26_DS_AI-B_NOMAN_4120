```python
#Z-score
import numpy as np
data=[10,20,30,40,50]
mean=np.mean(data)
print("Mean:",mean)
std_dev=np.std(data,ddof=0)
print("Standard Deviation:",std_dev)
x=40
z_score=(x-mean)/std_dev
print("Z-score:",z_score)
z_scores=[(x-mean)/std_dev for x in data]
print("Z-scores:",z_scores)
```
